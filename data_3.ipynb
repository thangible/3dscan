```python
import pandas as pd
import numpy as np
import os
from datetime import datetime

# Base directory containing all specimen data folders
base_data_dir = r'..\data\specimen_data'

# Find all directories in the base data directory that contain specimen data
specimen_data_dirs = []
if os.path.exists(base_data_dir):
    for item in os.listdir(base_data_dir):
        item_path = os.path.join(base_data_dir, item)
        if os.path.isdir(item_path):
            # Check if this directory contains the required CSV files
            required_files = ['specimens.csv', 'tensile_specimens.csv', 'polish_specimens.csv', 'notch_specimens.csv']
            if all(os.path.exists(os.path.join(item_path, file)) for file in required_files):
                specimen_data_dirs.append(item_path)

print(f"Found {len(specimen_data_dirs)} specimen data directories:")
for dir_path in specimen_data_dirs:
    print(f"  {os.path.basename(dir_path)}")

# Function to extract just the filename from URL paths
def extract_filename(url_string):
    """
    Extract the last filename from a URL string that may contain multiple paths separated by semicolons
    Example: "/documents/generic-form/.../RE-C015-S9_9bf94606_.bmp;documents/generic-form/.../RE-C015-S9.bmp" 
    Returns: "RE-C015-S9.bmp"
    """
    if pd.isna(url_string) or url_string == '':
        return url_string
    
    # Split by semicolon and take the last part
    parts = str(url_string).split(';')
    last_part = parts[-1]
    
    # Extract just the filename (everything after the last /)
    filename = last_part.split('/')[-1]
    
    return filename

# ...existing code...

# Function to analyze porosity statistics
def analyze_porosity_statistics(complete_dataset):
    """Analyze porosity statistics across all available porosity columns"""
    
    # Define porosity columns to check
    porosity_columns = [
        'porosity_3a_polish',
        'porosity_1_polish',
        'porosity_2_polish',
        'porosity_3_polish',
        'porosity_algorithm_polish',
        'porosity_polish'  # Adding the original porosity column as fallback
    ]
    
    # Find available porosity columns
    available_porosity_cols = [col for col in porosity_columns if col in complete_dataset.columns]
    
    if not available_porosity_cols:
        return {
            'columns_found': 0,
            'total_measurements': 0,
            'combined_stats': None,
            'per_column_stats': [],
            'message': 'No porosity columns found'
        }
    
    # Per-column statistics
    per_column_stats = []
    all_values = []
    
    for col in available_porosity_cols:
        # Convert to numeric, coercing errors to NaN
        numeric_values = pd.to_numeric(complete_dataset[col], errors='coerce')
        non_na_count = numeric_values.notna().sum()
        total_count = len(complete_dataset)
        percentage = (non_na_count / total_count * 100) if total_count > 0 else 0
        
        per_column_stats.append({
            'column': col,
            'non_null_count': non_na_count,
            'total_count': total_count,
            'percentage': round(percentage, 2)
        })
        
        # Collect non-null values for combined statistics
        if non_na_count > 0:
            all_values.append(numeric_values.dropna())
    
    # Combined statistics across all porosity columns
    combined_stats = None
    total_measurements = 0
    
    if all_values:
        combined_porosity = pd.concat(all_values, ignore_index=True)
        total_measurements = len(combined_porosity)
        
        combined_stats = {
            'count': total_measurements,
            'mean': round(combined_porosity.mean(), 4),
            'std': round(combined_porosity.std(), 4),
            'min': round(combined_porosity.min(), 4),
            'q25': round(combined_porosity.quantile(0.25), 4),
            'median': round(combined_porosity.median(), 4),
            'q75': round(combined_porosity.quantile(0.75), 4),
            'max': round(combined_porosity.max(), 4)
        }
    
    return {
        'columns_found': len(available_porosity_cols),
        'available_columns': available_porosity_cols,
        'total_measurements': total_measurements,
        'combined_stats': combined_stats,
        'per_column_stats': per_column_stats
    }

# Function to calculate specimens with both images and porosity data
def calculate_images_and_porosity_overlap(complete_dataset, available_image_columns, available_porosity_cols):
    """Calculate the number of specimens that have both images and porosity data"""
    
    if not available_image_columns or not available_porosity_cols:
        return 0
    
    # Check which specimens have at least one image URL
    has_images_mask = complete_dataset[available_image_columns].notna().any(axis=1)
    
    # Check which specimens have at least one porosity value
    has_porosity_mask = pd.Series(False, index=complete_dataset.index)
    for col in available_porosity_cols:
        if col in complete_dataset.columns:
            numeric_values = pd.to_numeric(complete_dataset[col], errors='coerce')
            has_porosity_mask |= numeric_values.notna()
    
    # Count specimens with both images and porosity
    has_both = has_images_mask & has_porosity_mask
    
    return has_both.sum()

# Function to process a single data directory
def process_data_directory(data_dir):
    """Process a single specimen data directory and return analysis results"""
    
    print(f"\n{'='*80}")
    print(f"PROCESSING: {os.path.basename(data_dir)}")
    print(f"{'='*80}")
    
    try:
        # Read the CSV files
        specimens_path = os.path.join(data_dir, 'specimens.csv')
        tensile_path = os.path.join(data_dir, 'tensile_specimens.csv')
        polish_path = os.path.join(data_dir, 'polish_specimens.csv')
        notch_path = os.path.join(data_dir, 'notch_specimens.csv')

        specimens_df = pd.read_csv(specimens_path, sep=';')
        tensile_df = pd.read_csv(tensile_path, sep=';')
        polish_df = pd.read_csv(polish_path, sep=';')
        notch_df = pd.read_csv(notch_path, sep=';')

        # Display basic information about each dataframe
        print("=== SPECIMENS (Main) ===")
        print(f"Shape: {specimens_df.shape}")
        print(f"Columns: {list(specimens_df.columns)}")
        print("\n=== TENSILE SPECIMENS ===")
        print(f"Shape: {tensile_df.shape}")
        print(f"Columns: {list(tensile_df.columns)}")
        print("\n=== POLISH SPECIMENS ===")
        print(f"Shape: {polish_df.shape}")
        print(f"Columns: {list(polish_df.columns)}")
        print("\n=== NOTCH SPECIMENS ===")
        print(f"Shape: {notch_df.shape}")
        print(f"Columns: {list(notch_df.columns)}")

        # Function to link all tables together
        def create_linked_dataset():
            """Link all specimen tables together using the key relationships"""
            
            # Start with the main specimens table
            linked_data = specimens_df.copy()
            
            # Add tensile test data
            if not tensile_df.empty:
                tensile_data = tensile_df.add_suffix('_tensile').rename(columns={'specimens_key_tensile': 'key'})
                linked_data = linked_data.merge(tensile_data, on='key', how='left')
            
            # Add polish test data
            if not polish_df.empty:
                polish_data = polish_df.add_suffix('_polish').rename(columns={'specimens_key_polish': 'key'})
                linked_data = linked_data.merge(polish_data, on='key', how='left')
            
            # Add notch test data
            if not notch_df.empty:
                notch_data = notch_df.add_suffix('_notch').rename(columns={'specimens_key_notch': 'key'})
                linked_data = linked_data.merge(notch_data, on='key', how='left')
            
            return linked_data

        # Create the linked dataset
        complete_dataset = create_linked_dataset()

        print(f"\nComplete dataset shape: {complete_dataset.shape}")
        print(f"Original specimens: {len(specimens_df)}")
        print(f"With tensile data: {complete_dataset.get('name_tensile', pd.Series()).notna().sum()}")
        print(f"With polish data: {complete_dataset.get('name_polish', pd.Series()).notna().sum()}")
        print(f"With notch data: {complete_dataset.get('name_notch', pd.Series()).notna().sum()}")

        # Analyze porosity statistics
        porosity_analysis = analyze_porosity_statistics(complete_dataset)
        
        print(f"\n=== POROSITY ANALYSIS ===")
        print(f"Porosity columns found: {porosity_analysis['columns_found']}")
        print(f"Total porosity measurements: {porosity_analysis['total_measurements']}")
        
        if porosity_analysis['combined_stats']:
            stats = porosity_analysis['combined_stats']
            print(f"Porosity statistics (combined):")
            print(f"  Mean: {stats['mean']}")
            print(f"  Median: {stats['median']}")
            print(f"  Range: {stats['min']} - {stats['max']}")
            print(f"  Q25-Q75: {stats['q25']} - {stats['q75']}")

        # Define the image URL columns
        image_url_columns = [
            'polish_img_url_3a_polish',
            'polish_img_url_6a_polish', 
            'polish_img_url_1_orig_polish',
            'polish_img_url_2_orig_polish',
            'polish_img_url_3_orig_polish',
            'polish_img_url_4_orig_polish',
            'polish_img_url_5_orig_polish',
            'polish_img_url_6_orig_polish',
            'polish_img_url_3a_orig_polish',
            'polish_img_url_6a_orig_polish',
            'polish_img_url_1_polish',
            'polish_img_url_2_polish',
            'polish_img_url_3_polish',
            'polish_img_url_4_polish',
            'polish_img_url_5_polish',
            'polish_img_url_6_polish'
        ]

        # Find available image URL columns
        available_image_columns = [col for col in image_url_columns if col in complete_dataset.columns]
        
        specimens_with_images = pd.DataFrame()
        if available_image_columns:
            # Filter rows where at least one image URL column is not null
            has_images_mask = complete_dataset[available_image_columns].notna().any(axis=1)
            
            # Get rows with context (DataFrame) - only specimens that have at least one image
            specimens_with_images = complete_dataset.loc[
                has_images_mask,
                ['key'] + available_image_columns
            ].copy()
            
            # Apply filename extraction
            for col in available_image_columns:
                specimens_with_images[col] = specimens_with_images[col].apply(extract_filename)

        print(f"Found {len(specimens_with_images)} specimens with at least one image URL")

        # Calculate specimens with both images and porosity data
        specimens_with_images_and_porosity = calculate_images_and_porosity_overlap(
            complete_dataset, 
            available_image_columns, 
            porosity_analysis.get('available_columns', [])
        )
        
        print(f"Found {specimens_with_images_and_porosity} specimens with both images and porosity data")

        # Image file analysis
        images_dir = os.path.join(data_dir, 'images')
        image_analysis_results = {}
        files_not_in_db_examples = []
        db_not_in_files_examples = []
        
        if os.path.exists(images_dir):
            # Get all files in the images directory
            image_files = [f for f in os.listdir(images_dir) if os.path.isfile(os.path.join(images_dir, f))]
            print(f"Found {len(image_files)} files in images directory")
            
            # Convert to set for faster lookup
            image_files_set = set(image_files)
            
            # Get all filenames from specimens_with_images (after cleaning)
            database_filenames = set()
            for col in available_image_columns:
                if col in specimens_with_images.columns:
                    # Get non-null values and add to set
                    non_null_values = specimens_with_images[col].dropna()
                    database_filenames.update(non_null_values)
            
            print(f"Found {len(database_filenames)} unique filenames in database")
            
            # Find matches
            matches = image_files_set.intersection(database_filenames)
            print(f"Found {len(matches)} matching files")
            
            # Find files in directory but not in database
            files_not_in_db = image_files_set - database_filenames
            print(f"Files in directory but not in database: {len(files_not_in_db)}")
            
            # Find filenames in database but not in directory
            db_not_in_files = database_filenames - image_files_set
            print(f"Database entries not found in directory: {len(db_not_in_files)}")
            
            # Get examples for non-matching files
            files_not_in_db_examples = sorted(list(files_not_in_db))[:10]
            db_not_in_files_examples = sorted(list(db_not_in_files))[:10]
            
            image_analysis_results = {
                'total_files': len(image_files),
                'database_filenames': len(database_filenames),
                'matches': len(matches),
                'files_not_in_db': len(files_not_in_db),
                'db_not_in_files': len(db_not_in_files),
                'files_not_in_db_examples': files_not_in_db_examples,
                'db_not_in_files_examples': db_not_in_files_examples
            }
        else:
            print(f"Images directory not found: {images_dir}")
            image_analysis_results = {
                'total_files': 0,
                'database_filenames': 0,
                'matches': 0,
                'files_not_in_db': 0,
                'db_not_in_files': 0,
                'directory_missing': True,
                'files_not_in_db_examples': [],
                'db_not_in_files_examples': []
            }

        # Data summary
        print(f"\n=== DATA SUMMARY for {os.path.basename(data_dir)} ===")

        # Count specimens by material
        print("\nSpecimens by Material:")
        if 'material' in specimens_df.columns:
            print(specimens_df['material'].value_counts())
        else:
            print("Material column not found")

        # Count specimens by customer
        print("\nTop 10 Customers:")
        if 'customer_name' in specimens_df.columns:
            print(specimens_df['customer_name'].value_counts().head(10))
        else:
            print("Customer name column not found")

        # Return summary data for this directory
        return {
            'directory': os.path.basename(data_dir),
            'specimens_count': len(specimens_df),
            'tensile_count': len(tensile_df),
            'polish_count': len(polish_df),
            'notch_count': len(notch_df),
            'complete_dataset_shape': complete_dataset.shape,
            'specimens_with_images': len(specimens_with_images),
            'specimens_with_images_and_porosity': specimens_with_images_and_porosity,
            'image_analysis': image_analysis_results,
            'materials_count': specimens_df['material'].nunique() if 'material' in specimens_df.columns else 0,
            'customers_count': specimens_df['customer_name'].nunique() if 'customer_name' in specimens_df.columns else 0,
            'porosity_analysis': porosity_analysis
        }
        
    except Exception as e:
        print(f"Error processing {data_dir}: {str(e)}")
        return {
            'directory': os.path.basename(data_dir),
            'error': str(e)
        }

# Process all directories
all_results = []
for data_dir in specimen_data_dirs:
    result = process_data_directory(data_dir)
    all_results.append(result)

# Create overall summary report
print(f"\n{'='*80}")
print("OVERALL SUMMARY")
print(f"{'='*80}")

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
summary_filename = f"specimen_data_summary.txt"

with open(summary_filename, 'w', encoding='utf-8') as f:
    f.write("=" * 80 + "\n")
    f.write("SPECIMEN DATA DIRECTORIES ANALYSIS SUMMARY\n")
    f.write("=" * 80 + "\n")
    f.write(f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
    f.write(f"Total directories processed: {len(all_results)}\n\n")
    
    # Summary table
    f.write("DIRECTORY SUMMARY:\n")
    f.write("-" * 90 + "\n")
    f.write(f"{'Directory':<30} {'Specimens':<10} {'Images':<8} {'Porosity':<10} {'Both':<8} {'Materials':<10}\n")
    f.write("-" * 90 + "\n")
    
    total_specimens = 0
    total_images = 0
    total_porosity = 0
    total_both = 0
    
    for result in all_results:
        if 'error' not in result:
            specimens = result['specimens_count']
            images = result['specimens_with_images']
            materials = result['materials_count']
            porosity_count = result.get('porosity_analysis', {}).get('total_measurements', 0)
            both_count = result.get('specimens_with_images_and_porosity', 0)
            
            total_specimens += specimens
            total_images += images
            total_porosity += porosity_count
            total_both += both_count
            
            f.write(f"{result['directory']:<30} {specimens:<10} {images:<8} {porosity_count:<10} {both_count:<8} {materials:<10}\n")
        else:
            f.write(f"{result['directory']:<30} {'ERROR':<10} {'ERROR':<8} {'ERROR':<10} {'ERROR':<8} {'ERROR':<10}\n")
    
    f.write("-" * 90 + "\n")
    f.write(f"{'TOTAL':<30} {total_specimens:<10} {total_images:<8} {total_porosity:<10} {total_both:<8} {'':<10}\n")
    
    # Detailed results for each directory
    f.write("\n\nDETAILED RESULTS:\n")
    f.write("=" * 80 + "\n")
    
    for result in all_results:
        f.write(f"\nDirectory: {result['directory']}\n")
        f.write("-" * 40 + "\n")
        
        if 'error' in result:
            f.write(f"ERROR: {result['error']}\n")
        else:
            f.write(f"Specimens: {result['specimens_count']}\n")
            f.write(f"Tensile records: {result['tensile_count']}\n")
            f.write(f"Polish records: {result['polish_count']}\n")
            f.write(f"Notch records: {result['notch_count']}\n")
            f.write(f"Complete dataset shape: {result['complete_dataset_shape']}\n")
            f.write(f"Specimens with images: {result['specimens_with_images']}\n")
            f.write(f"Specimens with images AND porosity: {result.get('specimens_with_images_and_porosity', 0)}\n")
            f.write(f"Unique materials: {result['materials_count']}\n")
            f.write(f"Unique customers: {result['customers_count']}\n")
            
            # Porosity analysis
            porosity_data = result.get('porosity_analysis', {})
            f.write(f"\nPOROSITY ANALYSIS:\n")
            f.write(f"Porosity columns found: {porosity_data.get('columns_found', 0)}\n")
            f.write(f"Total porosity measurements: {porosity_data.get('total_measurements', 0)}\n")
            
            if porosity_data.get('combined_stats'):
                stats = porosity_data['combined_stats']
                f.write(f"Porosity statistics (combined across all columns):\n")
                f.write(f"  Count: {stats['count']}\n")
                f.write(f"  Mean: {stats['mean']}\n")
                f.write(f"  Std Dev: {stats['std']}\n")
                f.write(f"  Min: {stats['min']}\n")
                f.write(f"  Q25: {stats['q25']}\n")
                f.write(f"  Median: {stats['median']}\n")
                f.write(f"  Q75: {stats['q75']}\n")
                f.write(f"  Max: {stats['max']}\n")
            
            if porosity_data.get('per_column_stats'):
                f.write(f"\nPer-column porosity statistics:\n")
                for col_stat in porosity_data['per_column_stats']:
                    f.write(f"  {col_stat['column']}: {col_stat['non_null_count']}/{col_stat['total_count']} ({col_stat['percentage']}%)\n")
            
            # Image analysis
            img_analysis = result['image_analysis']
            f.write(f"\nIMAGE ANALYSIS:\n")
            if 'directory_missing' not in img_analysis:
                f.write(f"Image files in directory: {img_analysis['total_files']}\n")
                f.write(f"Matching image files: {img_analysis['matches']}\n")
                f.write(f"Files not in DB: {img_analysis['files_not_in_db']}\n")
                f.write(f"DB entries not found: {img_analysis['db_not_in_files']}\n")
                
                # Add examples of non-matching files
                f.write("\nExamples of files in directory but not in database (max 10):\n")
                for i, filename in enumerate(img_analysis['files_not_in_db_examples']):
                    f.write(f"  {i+1}. {filename}\n")
                
                f.write("\nExamples of database entries not found in directory (max 10):\n")
                for i, filename in enumerate(img_analysis['db_not_in_files_examples']):
                    f.write(f"  {i+1}. {filename}\n")
            else:
                f.write("Images directory: NOT FOUND\n")

print(f"\nOverall summary saved to: {summary_filename}")
print(f"Processed {len(all_results)} directories")
print(f"Total specimens across all directories: {sum(r.get('specimens_count', 0) for r in all_results if 'error' not in r)}")
print(f"Total specimens with images: {sum(r.get('specimens_with_images', 0) for r in all_results if 'error' not in r)}")
print(f"Total porosity measurements: {sum(r.get('porosity_analysis', {}).get('total_measurements', 0) for r in all_results if 'error' not in r)}")
print(f"Total specimens with both images and porosity: {sum(r.get('specimens_with_images_and_porosity', 0) for r in all_results if 'error' not in r)}")
```
