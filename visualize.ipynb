```python
from dash import Dash, dcc, html, Input, Output, no_update, callback
import plotly.graph_objects as go
import pandas as pd
import requests
import os

url = "https://raw.githubusercontent.com/plotly/dash-sample-apps/main/apps/dash-drug-discovery/data/small_molecule_drugbank.csv"
response = requests.get(url)
data_path = 'datasets/small_molecule_drugbank.csv'
os.makedirs(os.path.dirname(data_path), exist_ok=True)
with open(data_path, "wb") as f:
    f.write(response.content)
# Small molecule drugbank dataset
# Source: https://raw.githubusercontent.com/plotly/dash-sample-apps/main/apps/dash-drug-discovery/data/small_molecule_drugbank.csv'


df = pd.read_csv(data_path, header=0, index_col=0)

fig = go.Figure(data=[
    go.Scatter(
        x=df["LOGP"],
        y=df["PKA"],
        mode="markers",
        marker=dict(
            colorscale='viridis',
            color=df["MW"],
            size=df["MW"],
            colorbar={"title": "Molecular<br>Weight"},
            line={"color": "#444"},
            reversescale=True,
            sizeref=45,
            sizemode="diameter",
            opacity=0.8,
        )
    )
])

# turn off native plotly.js hover effects - make sure to use
# hoverinfo="none" rather than "skip" which also halts events.
fig.update_traces(hoverinfo="none", hovertemplate=None)

fig.update_layout(
    xaxis=dict(title='Log P'),
    yaxis=dict(title='pkA'),
    plot_bgcolor='rgba(255,255,255,0.1)'
)

app = Dash()

app.layout = html.Div([
    dcc.Graph(id="graph-basic-2", figure=fig, clear_on_unhover=True),
    dcc.Tooltip(id="graph-tooltip"),
])


@callback(
    Output("graph-tooltip", "show"),
    Output("graph-tooltip", "bbox"),
    Output("graph-tooltip", "children"),
    Input("graph-basic-2", "hoverData"),
)
def display_hover(hoverData):
    if hoverData is None:
        return False, no_update, no_update

    # demo only shows the first point, but other points may also be available
    pt = hoverData["points"][0]
    bbox = pt["bbox"]
    num = pt["pointNumber"]

    df_row = df.iloc[num]
    img_src = df_row['IMG_URL']
    name = df_row['NAME']
    form = df_row['FORM']
    desc = df_row['DESC']
    if len(desc) > 300:
        desc = desc[:100] + '...'

    children = [
        html.Div([
            html.Img(src=img_src, style={"width": "100%"}),
            html.H2(f"{name}", style={"color": "darkblue", "overflow-wrap": "break-word"}),
            html.P(f"{form}"),
            html.P(f"{desc}"),
        ], style={'width': '200px', 'white-space': 'normal'})
    ]

    return True, bbox, children


if __name__ == "__main__":
    app.run(debug=True)
```

```python
import torch
import os

def inspect_checkpoint(checkpoint_path):
    """Inspect a PyTorch checkpoint file"""
    print(f"Inspecting checkpoint: {checkpoint_path}")
    print("=" * 60)
    
    # Load the checkpoint
    checkpoint = torch.load(checkpoint_path, map_location='cpu')
    
    # Check the type and structure
    print(f"Checkpoint type: {type(checkpoint)}")
    
    if isinstance(checkpoint, dict):
        print(f"Checkpoint keys: {list(checkpoint.keys())}")
        print()
        
        # If it contains metadata
        if 'model_state_dict' in checkpoint:
            print("This checkpoint contains metadata:")
            for key in checkpoint.keys():
                if key != 'model_state_dict':
                    print(f"  {key}: {checkpoint[key]}")
            print()
            
            # Inspect the actual model parameters
            state_dict = checkpoint['model_state_dict']
        else:
            # Direct state dict
            state_dict = checkpoint
            print("This checkpoint contains direct model state dict")
    else:
        # Old format - direct state dict
        state_dict = checkpoint
        print("This checkpoint contains direct model parameters")
    
    print("\nModel Parameters:")
    print("-" * 40)
    
    total_params = 0
    for name, param in state_dict.items():
        param_count = param.numel()
        total_params += param_count
        print(f"{name:40} | Shape: {str(param.shape):20} | Params: {param_count:8}")
    
    print("-" * 40)
    print(f"Total parameters: {total_params:,}")
    
    return state_dict

# Usage
checkpoint_path = "exp/best_vqvae_model.pth"
if os.path.exists(checkpoint_path):
    state_dict = inspect_checkpoint(checkpoint_path)
else:
    print(f"Checkpoint not found: {checkpoint_path}")
```

```python
from model.vq_vae import VQVAE

def compare_model_with_checkpoint(checkpoint_path, model_params):
    """Compare model architecture with saved checkpoint"""
    
    # Create model with your parameters
    model = VQVAE(**model_params)
    model_state = model.state_dict()
    
    # Load checkpoint
    checkpoint = torch.load(checkpoint_path, map_location='cpu')
    if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
        saved_state = checkpoint['model_state_dict']
    else:
        saved_state = checkpoint
    
    print("COMPARISON RESULTS:")
    print("=" * 60)
    
    model_keys = set(model_state.keys())
    saved_keys = set(saved_state.keys())
    
    print(f"Model parameters: {len(model_keys)}")
    print(f"Saved parameters: {len(saved_keys)}")
    
    # Missing in model
    missing_in_model = saved_keys - model_keys
    if missing_in_model:
        print(f"\nMissing in current model ({len(missing_in_model)}):")
        for key in sorted(missing_in_model):
            print(f"  {key}")
    
    # Missing in checkpoint
    missing_in_checkpoint = model_keys - saved_keys
    if missing_in_checkpoint:
        print(f"\nMissing in checkpoint ({len(missing_in_checkpoint)}):")
        for key in sorted(missing_in_checkpoint):
            print(f"  {key}")
    
    # Shape mismatches
    common_keys = model_keys & saved_keys
    shape_mismatches = []
    for key in common_keys:
        if model_state[key].shape != saved_state[key].shape:
            shape_mismatches.append((key, model_state[key].shape, saved_state[key].shape))
    
    if shape_mismatches:
        print(f"\nShape mismatches ({len(shape_mismatches)}):")
        for key, model_shape, saved_shape in shape_mismatches:
            print(f"  {key}: model{model_shape} vs saved{saved_shape}")
    
    if not missing_in_model and not missing_in_checkpoint and not shape_mismatches:
        print("\n✅ Perfect match! Model and checkpoint are compatible.")
    
    return len(missing_in_model) == 0 and len(missing_in_checkpoint) == 0 and len(shape_mismatches) == 0

# Usage example
model_params = {
    'in_channels': 1,
    'out': 1,
    'embedding_dim': 512,  # This should match your config
    'num_embeddings': 512,  # This should match your config
    'hidden_dims': [128, 256],  # Default from your model
    'beta': 0.25,
    'img_size': 64
}

is_compatible = compare_model_with_checkpoint("exp/best_vqvae_model.pth", model_params)
```
